Test that Python, packages, and the Census API key work

```python
import os
import requests
from dotenv import load_dotenv

load_dotenv()
key = os.getenv("CENSUS_API_KEY")

url = "https://api.census.gov/data/2024/acs/acs5"
params = {"get": "NAME,B19013_001E", "for": "state:24", "key": key}

response = requests.get(url, params=params)
print(response.json())
```
