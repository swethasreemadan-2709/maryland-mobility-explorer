# 03 – Load DuckDB
Loads the clean county indicators table into a local DuckDB database for SQL analysis.

```python
import duckdb
import pandas as pd

indicators = pd.read_csv("../data/processed/county_indicators.csv", dtype={"fips": str})

con = duckdb.connect("../data/processed/mobility.duckdb")
con.execute("CREATE OR REPLACE TABLE county_indicators AS SELECT * FROM indicators")
```

```python
con.sql("""
    SELECT year, COUNT(*) AS counties
    FROM county_indicators
    GROUP BY year
""").show()
```

```python
con.sql("""
    SELECT county_name, poverty_rate, rent_burden_rate
    FROM county_indicators
    WHERE year = 2024
    ORDER BY poverty_rate DESC
    LIMIT 5
""").show()
```

```python
con.close()
```
